#  Q10 End-to-End Data Wrangling Task

Perform the following workflow:
    
* Filter the dataset where Sales > 3000

* Group data by Region and Product

* Calculate total sales

* Reshape the result into a pivot table

Explain the business insights obtained from this analysis

```python
filtered = df[df['Sales'] > 3000]
filtered_group = filtered.groupby(['Region','Product'])['Sales'].sum().reset_index()
pivot = filtered_group.pivot(index = 'Region', columns = 'Product', values='Sales')
pivot['total_sales'] = pivot.sum(axis=1)
pivot.loc ['Grand_total'] =  pivot.sum(axis=0)
print(pivot)
```

# Interpretation 

* East region has highest sale & Nort region has lowest sale
* phone is high revenue genrator
